### DataFrame = A tabular data structure with rows AND columns. (2 Dimensional)
#### Similar to an Excel spreadsheet

```python
import pandas as pd

data = {"Name": ["Alex", "Robert", "Mars", "Patrick"],
        "Age": [34, 38, 29, 43]}

df = pd.DataFrame(data, index=[101, 102, 103, 104])
```

```python
df
```

#### to access a single row we can use either loc or iloc property

```python
print(df.loc[103])
print(df.loc[101])
```

```python
print(df.iloc[3])
print(df.iloc[-1])
```

#### to add a new column

```python
df["Job"] = ["Developer", "Designer", "N/A", "Intern"]
df["Salary"] = [65000, 45000, 0, 36000]
```

```python
df
```

#### to add a new row, we have to create a new data frame and concat with original data frame

```python
new_row = pd.DataFrame([{"Name": "Zen", "Age": 46, "Job": "Sales", "Salary": 79000}, {"Name": "Ross", "Age": 24, "Job": "Marketing", "Salary": 18000} ], index=[105, 106])
```

```python
df = pd.concat([df, new_row])
```
