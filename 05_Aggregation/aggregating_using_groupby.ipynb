### Used to summarize and analyze data, Often used with the groupby()I function


```python
import pandas as pd

df = pd.read_csv("../all_files/data.csv")
print(df)
```

```python
group = df.groupby("Type1")
```

```python
print(group["Height"].mean())
# it means grouping the Type1 and for that category we are calculating avg of height
```

```python
# to get the min height from each type

print(group["Height"].min())
```
