### Filtering - keeping the rows that match the condition

```python
import pandas as pd

df = pd.read_csv("../all_files/data.csv")
```

```python
print(df)
```

```python
tall_pokemon = df["Height"] >= 2
df[tall_pokemon]
```

```python
heavy_pokemon = df["Weight"] > 100
df[heavy_pokemon]
```

```python
water_pokemon = (df["Type1"] == "Water") | (df["Type2"] == "Water")
df[water_pokemon]
```

```python
fire_fly_pokemon = (df["Type1"] == "Fire") & (df["Type2"] == "Flying")
df[fire_fly_pokemon]
```

```python
# legendary_pokemon = df["Legendary"] == 1
legendary_pokemon = df["Legendary"] == True
df[legendary_pokemon]
```
