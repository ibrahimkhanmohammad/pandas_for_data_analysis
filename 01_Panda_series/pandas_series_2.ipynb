```python
import pandas as pd

calories = {"Day 1": 1950, "Day 2": 2290, "Day 3": 1520}

calory_series = pd.Series(calories, name = "Calories")
```

```python
calory_series
```

### accessing the values is done by iloc method

```python
print(calory_series.iloc[1])
print(calory_series.iloc[0])
```

### accessing the indexes is done by loc method

```python
calory_series.loc["Day 2"]
```

```python
calory_series[calory_series >= 1500]
```

### filtering can be done easily as follows

```python
calory_series[(calory_series >= 1500) & (calory_series <=2000)]
```

### changing the values by index

```python
calory_series.loc["Day 3"] += 500
```
