### Series = A Pandas 1-Dimensional labeled array that can hold any data type

#### Think of it like a single column in a spreadsheet (1-Dimensional)

```python
import pandas as pd

rolls = [101, 103, 105, 107, 109]
names = ["alex", "bob", "carry", "delphi", "justin"]

roll_series = pd.Series(rolls, index=names, name="Rolls")
```

```python
roll_series
```

### accessing the values is done by iloc method

```python
print(roll_series.iloc[1])
print(roll_series.iloc[-1])
```

### accessing the indexes is done by loc method

```python
print(roll_series.loc["carry"])
print(roll_series.loc["delphi"])
```

### filtering can be done easily as follows

```python
roll_series[roll_series >= 105]
```

```python
roll_series[roll_series <= 104]
```

```python
roll_series[(roll_series >= 103) & (roll_series <= 108)]
```

### changing the values by index

```python
roll_series.loc["carry"] = 111
print(roll_series)
```
